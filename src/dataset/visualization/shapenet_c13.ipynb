```python
import matplotlib as mpl
# Set the Agg backend for headless plotting, MUST be done before importing pyplot
mpl.use('Agg')

import nrrd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Read the NRRD file
file_path = "../../../shapenet_colored.nrrd"
data, header = nrrd.read(file_path)

print(f"Data shape: {data.shape}")

# Separate the RGBA channels
# The shape is (4, 32, 32, 32) -> (channels, Z, Y, X)
red_channel = data[0, :, :, :]
green_channel = data[1, :, :, :]
blue_channel = data[2, :, :, :]
alpha_channel = data[3, :, :, :]

# Create a boolean array to identify solid voxels
solid_voxels = alpha_channel > 0

# Step 2: Prepare the color array for voxels()
# Matplotlib's voxels() expects a 4D array of shape (N, M, P, 3) for RGB colors
# where (N, M, P) is the same shape as the `solid_voxels` boolean array.
# We create this empty array and fill it with our normalized RGB values.
colors = np.zeros(solid_voxels.shape + (3,))
colors[:,:,:,0] = red_channel # Red channel
colors[:,:,:,1] = green_channel # Green channel
colors[:,:,:,2] = blue_channel # Blue channel

# Normalize the color values (typically from 0-255 to 0.0-1.0)
colors_normalized = colors / 255.0

# Step 3: Visualize the 3D shape with matplotlib
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(projection='3d')

# Pass both the boolean mask and the color array to ax.voxels()
ax.voxels(solid_voxels, facecolors=colors_normalized, edgecolor='none')

ax.set_title("3D Shape Visualization with Matplotlib")
plt.savefig("shapenet_to_nrrd.png")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# Use Agg backend for headless environments (optional)
mpl.use('Agg')

# Load the colored voxel grid (shape: 32x32x32x4)
voxels = np.load('../../../shapenet_colored_32x32x32.npy')

# Extract colors and occupancy mask
colors = voxels[:, :, :, 0:3]  # RGB in 0-255 range
alpha = voxels[:, :, :, 3]     # Alpha channel (0 or 255)
occupancy = alpha > 0          # Boolean mask for filled voxels

# Normalize RGB to [0,1] for matplotlib
colors_normalized = colors / 255.0

# Prepare figure and 3D axis
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(projection='3d')

# Plot voxels: use occupancy mask and color each voxel by its RGB color
ax.voxels(occupancy, facecolors=colors_normalized, edgecolor='none')

ax.set_title("Colored Voxel Grid Visualization")
plt.savefig('colored_voxel_visualization.png')
plt.show()
```

```python
import random
from datasets import load_dataset
from collections import Counter

# Load the ShapeNet-C13 dataset
ds = load_dataset("3dlg-hcvc/ShapeNet-C13")

# Count the number of captions for each model_id
train_model_ids = ds['train']['model_id']
model_counts = Counter(train_model_ids)

# Filter models that have at least 2 captions
valid_models = [m for m, c in model_counts.items() if c >= 2]

# Randomly select a model from the valid ones
selected_model = random.choice(valid_models)

# Get all entries (captions) for the selected model
entries = [entry for entry in ds['train'] if entry['model_id'] == selected_model]

# Print sample information
print(f"Model ID: {selected_model}")
print(f"Category (synset_id): {entries[0]['synset_id']}")
print(f"Number of captions: {len(entries)}")
print("Captions:")
for e in entries:
    print("-", e['description'])
```

```python
from collections import defaultdict
# Lấy split train
train_ds = ds['train']

model_counts = defaultdict(int)

for example in train_ds:
    model_id = example['model_id']
    model_counts[model_id] += 1

num_unique_models = len(model_counts)
print(f"Number of unique models in train: {num_unique_models}")

for i, (model_id, count) in enumerate(model_counts.items()):
    print(f"Model {model_id} has {count} samples")
    if i >= 10:
        break

total_samples = sum(model_counts.values())
print(f"Total train samples: {total_samples}")
```
